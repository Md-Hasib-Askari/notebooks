```python
import pandas as pd

# Load the provided training and test datasets
train_path = "data/DIA_trainingset_RDKit_descriptors.csv"
test_path = "data/DIA_testset_RDKit_descriptors.csv"

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Basic info: shape and first few columns
train_info = {
    "shape": train_df.shape,
    "columns": train_df.columns.tolist()[:15]  # preview first 15 columns
}
test_info = {
    "shape": test_df.shape,
    "columns": test_df.columns.tolist()[:15]
}

train_info, test_info
```

```python
import json, datetime, pytz

# Update tracker file with dataset ingestion
tracker_path = "data/InterDIA_ML_tracking.json"

with open(tracker_path, "r") as f:
    tracking = json.load(f)

tz = pytz.timezone("Asia/Dhaka")
now = datetime.datetime.now(tz).isoformat()

# Update dataset status
tracking["datasets"]["InterDIA"].update({
    "status": "imported",
    "train_shape": train_df.shape,
    "test_shape": test_df.shape,
    "n_features": train_df.shape[1] - 2,  # minus Label, SMILES
    "updated_at": now
})

# Mark action A1 as done
for action in tracking["next_actions"]:
    if action["id"] == "A1":
        action["status"] = "done"
        action["updated_at"] = now

# Save updated tracker
with open(tracker_path, "w") as f:
    json.dump(tracking, f, indent=2)

tracker_path
```
